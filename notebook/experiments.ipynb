```python
from langchain.prompts import PromptTemplate
```

```python
retriever=vectorstore.as_retriever()
```

```python
retriever.invoke("llama2 finetuning benchmark experiments.")
```

```python
prompt_template = """
        Answer the question based on the context provided below. 
        If the context does not contain sufficient information, respond with: 
        "I do not have enough information about this."

        Context: {context}

        Question: {question}

        Answer:"""
```

```python
prompt=PromptTemplate(
    template=prompt_template,
    input_variables=["context", "question"]
)
```

```python
prompt
```

```python
from langchain_core.output_parsers import StrOutputParser
```

```python
parser=StrOutputParser()
```

```python
def format_docs(docs):
    return "\n\n".join([doc.page_content for doc in docs])
```

```python
from langchain_core.runnables import RunnablePassthrough
```

```python
rag_chain = (
    {"context": retriever | format_docs, "question": RunnablePassthrough()}
    | prompt
    | model
    | StrOutputParser()
)
```

```python
answer = rag_chain.invoke("tell me about llm")
```

```python
pdf_dir = os.path.join(os.getcwd(), "multidoc")
pdf_files = [f for f in os.listdir(pdf_dir) if f.endswith(".pdf")]
```

```python
documents = []
for pdf_file in pdf_files:
    file_path = os.path.join(pdf_dir, pdf_file)
    loader = PyPDFLoader(file_path)
    documents.extend(loader.load())
```

```python
len(documents)
```

```python
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size = 500,
    chunk_overlap=150,
    length_function=len
)
```

```python
docs = text_splitter.split_documents(documents)
```

```python
len(docs)
```

```python
docs[0].page_content
```

```python
from langchain.vectorstores import FAISS
```

```python
vectorstore = FAISS.from_documents(docs, embedding_model)
```

```python


from datetime import datetime
from zoneinfo import ZoneInfo
import uuid

def generate_session_id(prefix="session"):
    ist = ZoneInfo("Asia/Kolkata")
    return f"{prefix}_{datetime.now(ist).strftime('%Y%m%d_%H%M%S')}_{uuid.uuid4().hex[:8]}"
```
